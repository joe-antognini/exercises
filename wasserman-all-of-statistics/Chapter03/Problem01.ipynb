```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def simulate_mean(n, n_trials=1_000_000):
    flips = np.random.choice([0.5, 2], replace=True, size=(n_trials, n))
    outcomes = np.prod(flips, axis=1)
    return np.mean(outcomes)


def calculate_mean(n):
    return (5 / 4)**n
```

```python
for n in range(1, 11):
    print(n)
    print('Analytic expected value:', calculate_mean(n))
    print('Simulated mean:', simulate_mean(n))
    print()
```

```python
Ns = np.arange(1, 21)
calculated_means = [calculate_mean(n) for n in Ns]
simulated_means = [simulate_mean(n) for n in Ns]
```

```python
plt.plot(Ns, calculated_means, c='b', label='Analytic')
plt.scatter(Ns, simulated_means, c='r', label='Simulated')
plt.xlabel('n')
plt.ylabel('Expected value')

plt.yscale('log')
plt.legend()
```
